## Python Function for Wind Condition


```python
def wind_condition(speed):
    return "Calm" if speed < 5 else "Moderate" if speed <= 20 else "Strong"

# Example usage with data from the Excel file
import pandas as pd

# Load the CSV file
data = pd.read_csv('/content/seasonal_weather_data_exam.csv')

# Apply the function to the Wind Speed column
data['Wind Condition'] = data['Wind Speed'].apply(wind_condition)

# Display a preview of results
data[['Date', 'Wind Speed', 'Wind Condition']].head()
```

## Python Function Explanation

The function wind_condition() takes a wind speed and categorizes it as:

"Calm" if less than 5 km/h,
"Moderate" if between 5 and 20 km/h,
"Strong" if over 20 km/h.
This function simplifies the process by providing consistent results across the dataset. Using apply(), it automatically applies the categorization to each wind speed entry in the data.

# **Question 2: Data Visualisation with Matplotlib**

## Python Code for Visualizations

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv('/content/seasonal_weather_data_exam.csv')

# 1. Bar Plot for Daily Precipitation Levels
plt.figure(figsize=(10, 5))
plt.bar(data['Date'], data['Precipitation'], color='skyblue')
plt.xticks(rotation=90)
plt.xlabel('Date')
plt.ylabel('Precipitation (mm)')
plt.title('Daily Precipitation Levels')
plt.show()

# 2. Pie Chart for Rain Categories
# Categorize the precipitation levels
no_rain = data[data['Precipitation'] == 0].shape[0]
light_rain = data[(data['Precipitation'] > 0) & (data['Precipitation'] < 5)].shape[0]
heavy_rain = data[data['Precipitation'] >= 5].shape[0]

# Pie chart data
labels = ['No Rain', 'Light Rain', 'Heavy Rain']
sizes = [no_rain, light_rain, heavy_rain]
colors = ['lightgrey', 'lightblue', 'cornflowerblue']

plt.figure(figsize=(6, 6))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, startangle=140)
plt.title('Proportion of Days by Rain Category')
plt.show()
```

## Explanation of the Code

Bar Plot for Daily Precipitation:
The bar plot shows daily precipitation over a month. Each bar represents the precipitation in millimeters for a specific day.

Pie Chart for Rain Categories:

Three categories are created:

"No Rain" for days with 0 mm precipitation.

"Light Rain" for days with precipitation less than 5 mm but greater than 0.

"Heavy Rain" for days with precipitation 5 mm or more.

The pie chart shows the percentage of days in each category, helping visualize the frequency of different rain levels.

Observed Weather Patterns
The bar plot shows how precipitation levels vary across the month.

The pie chart indicates whether most days were dry, had light rain, or experienced heavy rain. This gives a quick overview of rain patterns for the month.

# **Question 4: Loops and Exception Handling**

## Python Code for Monitoring Temperature Readings


```python
import pandas as pd

# Load the dataset
data = pd.read_csv('/content/seasonal_weather_data_exam.csv')

# Simulate monitoring for 12 hours using the first 12 temperature readings from the dataset
temperature_readings = data['Temperature'][:12]

# Define a function to monitor temperature with exception handling
def monitor_temperature(readings):
    for hour, temp in enumerate(readings, start=1):
        try:
            if temp < 0:
                raise ValueError(f"Alert! Temperature below 0°C at hour {hour}. Stopping monitoring.")
            print(f"Hour {hour}: Temperature is {temp}°C")
        except ValueError as e:
            print(e)
            break

# Run the monitoring function
monitor_temperature(temperature_readings)
```

# Explanation

Loop Through Temperature Readings:

The loop checks temperature readings each hour (up to 12 readings).

**Exception Handling:**

If a temperature reading drops below 0°C, a ValueError is raised with a message, stopping further monitoring. This is handled by the except block, which displays the alert and breaks the loop.

**Result:**

The program outputs the temperature each hour until a reading below 0°C is detected, at which point it stops monitoring.

**Reflection on Exception Handling**

The exception handling in this program allows for an immediate response to critical conditions. By raising an exception when the temperature falls below 0°C, the program can safely stop further readings and issue an alert. This approach improves program control flow, ensuring prompt responses to unusual temperature changes without continuing unnecessary processing.

# **Question 5: Lists and Dictionaries**

## Python Code for Rainfall Tracker


```python
# Load the dataset
data = pd.read_csv('/content/seasonal_weather_data_exam.csv')

# Split the Precipitation data into four sections to represent four cities
# Each city gets approximately an equal part of the dataset
city_data = {
    "New York": data['Precipitation'][:len(data)//4].tolist(),
    "London": data['Precipitation'][len(data)//4:len(data)//2].tolist(),
    "Tokyo": data['Precipitation'][len(data)//2:3*len(data)//4].tolist(),
    "Perth": data['Precipitation'][3*len(data)//4:].tolist()
}

# Function to display precipitation data for a specific city
def display_city_precipitation(city):
    if city in city_data:
        print(f"Precipitation data for {city}: {city_data[city]}")
    else:
        print(f"{city} not found in the data.")

# Function to calculate total precipitation for all cities
def total_precipitation():
    total = sum(sum(readings) for readings in city_data.values())
    print(f"Total precipitation for all cities: {total} mm")

# Example usage
display_city_precipitation("New York")  # Display data for New York
total_precipitation()  # Calculate and display the total precipitation
```

# Explanation

**Data Split:**

The cities are named as New York, London, Tokyo, and Perth to add global context to the data.

**Display Precipitation Data:**

The display_city_precipitation(city) function retrieves and displays precipitation data for a specified city.

**Calculate Total Precipitation:**

total_precipitation() sums up precipitation values from all cities to get a combined total.
